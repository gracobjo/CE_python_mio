# PARTE I — CONTROL DE FLUJO

El **control de flujo** decide *qué* se ejecuta y *cuántas veces*. En Python hay tres familias:

| Familia | Palabras clave | Pregunta que responde |
|---|---|---|
| Selección | `if`, `elif`, `else`, `match`, `case` | ¿Este camino o aquel? |
| Repetición condicional | `while` | ¿Seguir mientras se cumpla esto? |
| Recorrido | `for` | ¿Para cada elemento de este iterable? |

Sin control de flujo un programa solo podría ejecutarse de arriba abajo, una sola vez. Con él puedes validar datos, clasificar notas, recorrer un dataset o mantener un menú hasta que el usuario salga.

## Epígrafe 1 — `if` / `elif` / `else`

### 1.1 Conceptos clave

| Concepto | Definición |
|---|---|
| **Condición** | Expresión que Python interpreta como verdadera o falsa (truthiness). |
| **`if`** | Ejecuta un bloque solo si la condición es verdadera. |
| **`elif`** | «else if»: se evalúa **solo** si los `if`/`elif` anteriores fueron falsos. |
| **`else`** | Camino residual: se ejecuta si ninguna condición anterior se cumplió. |
| **Operadores de comparación** | `==`, `!=`, `<`, `<=`, `>`, `>=`, `is`, `is not`, `in`, `not in`. |
| **Operadores lógicos** | `and`, `or`, `not`. Se evalúan de izquierda a derecha, con *short-circuit*. |
| **Truthiness** | No solo `True`/`False`: `0`, `0.0`, `''`, `[]`, `{}`, `set()`, `None` son falsos. |
| **Expresión condicional** | Ternario: `valor_si if condicion else valor_no`. |
| **Indentación** | En Python **delimita el bloque**. No es decorativa. |

Sintaxis general:

```python
if condicion_1:
    # bloque 1
elif condicion_2:
    # bloque 2
else:
    # bloque residual
```

`elif` y `else` son opcionales. Puede haber **cero o varios** `elif`. Solo un `else`, y siempre el último.


### 1.2 Explicación razonada

#### Qué ocurre

Python evalúa la primera condición. Si es verdadera, ejecuta **solo** ese bloque y **salta el resto** de la cadena `if`/`elif`/`else`. Si es falsa, pasa al siguiente `elif`, y así sucesivamente. El `else` no tiene condición: es «ninguna de las anteriores».

Eso implica una consecuencia de examen: **el orden de las condiciones importa**. Si pones primero `nota >= 5` y después `nota >= 9`, un 9 nunca llegará a Sobresaliente.

#### Por qué ocurre

Un `if` no «pregunta todas las ramas». Encadena exclusiones. Por eso es la herramienta correcta cuando las categorías son **mutuamente excluyentes** y tienen **prioridad**.

#### Condiciones, comparación y lógica

```python
edad >= 18           # comparación
edad >= 18 and ciudad == "Valladolid"   # conjunción
not es_menor         # negación
```

`==` compara **valores**. `is` compara **identidad** (el mismo objeto). Para `None` se usa `is None`, no `== None`.

`and` y `or` no devuelven necesariamente `True`/`False`: devuelven el **operando decisivo**.

```python
0 or [] or "Ana"     # "Ana"
"Ana" and 0          # 0
```

#### Evaluación de izquierda a derecha y short-circuit

- `A and B`: si `A` es falso, **no se evalúa** `B`.
- `A or B`: si `A` es verdadero, **no se evalúa** `B`.

Esto no es un detalle menor: si `B` tiene efectos (una llamada, una división, un acceso a diccionario), puede no ejecutarse. Por eso se usa a menudo como guarda:

```python
if alumno is not None and alumno["nota"] >= 5:
    ...
```

Si `alumno` es `None`, **no** se evalúa `alumno["nota"]` y no hay `TypeError`.

#### Cuándo usar `if` / `elif` / `else`

- Rangos numéricos (notas, edades, temperaturas).
- Validaciones (`if n < 0: ...`).
- Combinaciones lógicas complejas.
- Cuando cada rama depende de **expresiones distintas**, no de un mismo valor discreto.

#### Cuándo NO usarlo

- Si comparas **un mismo valor** con literales fijos (`"suma"`, `"resta"`, `"salir"`), `match` suele ser más claro.
- Si anidas tres o cuatro `if` para clasificar, casi siempre puedes aplanar a `elif`.
- No uses `if condicion: return True else: return False`; basta `return condicion`.

#### Truthiness

Python pregunta «¿este objeto se considera verdadero?», no solo «¿es el booleano True?».

| Valor | Se considera |
|---|---|
| `False`, `None`, `0`, `0.0`, `0j` | falso |
| `''`, `[]`, `()`, `{}`, `set()` | falso |
| Cualquier otro objeto habitual | verdadero (incluida una lista `[0]` o el string `"0"`) |

Por eso `if notas:` significa «si hay alguna nota», y `if not nombre:` significa «si el nombre está vacío».

#### Expresión condicional (ternario)

```python
estado = "apto" if media >= 5 else "no apto"
```

Es una **expresión**: produce un valor. El `if` de sentencia **no** produce un valor, controla un bloque. No sustituyas un `if`/`elif`/`else` de cuatro ramas por ternarios anidados: se vuelve ilegible.

#### Condiciones anidadas

Anidar es legal y a veces necesario (una decisión dentro de otra). El error habitual no es anidar: es anidar **sin necesidad**, duplicando condiciones o dejando ramas inalcanzables.

#### Errores de lógica frente a errores de sintaxis

Un `IndentationError` se ve al instante. Un `if nota > 5` que deja el 5 como suspenso es un **error de lógica**: el programa funciona y el resultado está mal. En examen, esos son los más caros.


````python
# ── Práctica guiada 1 — if / elif / else ─────────────────────────────────────
# Predice cada print ANTES de ejecutar.

# 1) if simple y par/impar
n = 7
if n % 2 == 0:
    print("par")
else:
    print("impar")

# 2) clasificación de edades (el orden de los elif es deliberado)
edad = 17
if edad < 13:
    categoria = "niño"
elif edad < 18:
    categoria = "adolescente"
elif edad < 65:
    categoria = "adulto"
else:
    categoria = "anciano"
print("edad 17 →", categoria)

# 3) notas: ¿qué ocurre con un 9 si el orden está mal?
nota = 9
if nota >= 5:
    print("ORDEN MAL: un 9 entra aquí y nunca llega a sobresaliente")
elif nota >= 9:
    print("esta rama es inalcanzable para un 9")

# 4) orden correcto por rangos de mayor a más específico, o con intervalos cerrados
if nota < 5:
    print("suspenso")
elif nota < 7:
    print("aprobado")
elif nota < 9:
    print("notable")
else:
    print("sobresaliente")

# 5) temperaturas y validación
temp = -3
if temp < -10 or temp > 50:
    print("lectura dudosa")
elif temp < 0:
    print("bajo cero")
else:
    print("no gela")

# 6) truthiness
print("if [] →", bool([]))
print("if [0] →", bool([0]))
print("if '0' →", bool("0"))

# 7) short-circuit: g() NO se llama
def f():
    print("  se llama f()")
    return True

def g():
    print("  se llama g()")
    return False

print("resultado f() or g() =", f() or g())

# 8) and/or devuelven el operando decisivo, no necesariamente True/False
print("0 or [] or 'Ana' or 'Luis' →", 0 or [] or "Ana" or "Luis")

# 9) ternario
media = 6.2
print("apto" if media >= 5 else "no apto")

# 10) guarda con short-circuit (no explota si alumno es None)
alumno = None
if alumno is not None and alumno["nota"] >= 5:
    print("aprobado")
else:
    print("sin alumno o suspenso")
````

### 1.4 Errores frecuentes

**1. Usar asignación `=` en lugar de comparación `==`**

INCORRECTO:
```python
if nota = 5:
    ...
```

CORRECTO:
```python
if nota == 5:
    ...
```

`=` no es una comparación: es un error de sintaxis dentro de un `if`.

**2. Indentación incorrecta (el `else` se «pega» al `if` equivocado)**

INCORRECTO:
```python
if edad >= 18:
    if ciudad == "Valladolid":
        print("ok")
else:
    print("menor")   # este else pertenece al if de edad, no al de ciudad
```

Si querías el `else` para la ciudad, debe ir al mismo nivel que `if ciudad`.

**3. Rangos mal ordenados (ramas inalcanzables)**

INCORRECTO:
```python
if nota >= 5:
    return "Aprobado"
elif nota >= 9:
    return "Sobresaliente"   # nunca se alcanza
```

CORRECTO: pregunta primero el caso más restrictivo, o usa intervalos (`5 <= nota < 7`).

**4. Confundir `and`/`or` con el lenguaje natural**

«Edad entre 18 y 65» no es `edad == 18 or 65`. Es `18 <= edad <= 65` (encadenamiento de comparaciones, válido en Python) o `edad >= 18 and edad <= 65`.

**5. Tratar `"0"` o `[0]` como falsos**

`"0"` es una cadena no vacía → verdadera. `[0]` es una lista no vacía → verdadera. Falsos son `0`, `''` y `[]`.

**6. Ternario anidado ilegible**

INCORRECTO:
```python
t = "A" if n >= 9 else "B" if n >= 7 else "C" if n >= 5 else "D"
```

Si hay más de dos caminos, usa `if`/`elif`/`else`.

**7. `if lista:` no distingue «lista vacía» de `None`**

Si `None` y `[]` deben tratarse distinto, usa `is None` explícitamente.

## Epígrafe 2 — `match` / `case`

### 2.1 Conceptos clave

| Concepto | Definición |
|---|---|
| **`match`** | Selección por **patrones** sobre un valor (Python 3.10+). |
| **`case`** | Una rama del `match`. Si el patrón encaja, se ejecuta el bloque. |
| **`case _`** | Comodín: encaja con cualquier valor no capturado antes. Equivale al `else`. |
| **Patrón literal** | Compara igualdad: `case "suma":`. |
| **Patrón de captura** | Nombra partes: `case (x, y):`. |
| **Guard** | Condición extra: `case n if n > 0:`. El patrón encaja **y** la guarda es verdadera. |
| **Or-pattern** | `case "s" \| "suma":` (varias alternativas). |

```python
match valor:
    case patron_1:
        ...
    case patron_2 if guarda:
        ...
    case _:
        ...
```


### 2.2 Explicación razonada

#### Qué es `match`

`match` **no** es un `switch` de C copiado. Es *structural pattern matching*: describe la **forma** del dato (literal, tupla, lista, diccionario) y, si encaja, extrae piezas.

```python
comando = ("suma", 3, 4)
match comando:
    case ("suma", a, b):
        resultado = a + b
```

Aquí no solo preguntas «¿el primer elemento es `"suma"`?»: **desempaquetas** `a` y `b` si la tupla tiene esa forma.

#### `match` frente a `if` / `elif`

| Criterio | `if` / `elif` | `match` / `case` |
|---|---|---|
| Comparas **expresiones distintas** (`edad < 18`, `nota >= 5`) | Natural | Forzado |
| Comparas **un mismo valor** con formas/literales | Funciona, más verboso | Más claro |
| Rangos numéricos | Ideal | Posible con guards, menos natural |
| Tuplas/listas/dicts con estructura | Manual | Idiomático |
| Python < 3.10 | Única opción | No existe |

**Cuándo usarlo:** menús, opcodes, comandos (`"suma"`, `"resta"`), clasificación de estructuras (`(0, y)`, `(x, 0)`), respuestas de APIs con forma fija.

**Cuándo NO:** clasificar una nota 0–10; validar `if n is None`; combinaciones lógicas heterogéneas. Ahí `if` es más honesto.

#### `case _`

`_` no guarda el valor. Es «cualquier otra cosa». Si omites el comodín y nada encaja, el `match` no hace nada (no lanza error).

#### Guards

```python
match n:
    case n if n > 0:
        signo = "positivo"
    case n if n < 0:
        signo = "negativo"
    case _:
        signo = "cero"
```

El patrón captura `n` y **después** se evalúa la guarda. Si la guarda falla, se prueba el siguiente `case`.

Cuidado: `case n:` (sin guarda) captura **todo** y deja inalcanzables los `case` siguientes. Es el equivalente a poner un `else` demasiado pronto.

#### Patrones de secuencia y de mapeo

```python
match punto:
    case (0, 0):
        ...
    case (0, y):
        ...
    case [a, b]:
        ...          # lista (o secuencia) de dos elementos

match alumno:
    case {"nombre": str() as nombre, "nota": nota} if nota >= 5:
        ...
```

No hace falta memorizar todos los patrones avanzados. Sí debes saber: **el orden de los `case` importa**, igual que el de los `elif`.


````python
# ── Práctica guiada 2 — match / case ─────────────────────────────────────────
# Calculadora por comando (valor discreto → match es más claro que if)

def calcular(operacion, a, b):
    match operacion:
        case "suma" | "s":
            return a + b
        case "resta" | "r":
            return a - b
        case "multiplicacion" | "m":
            return a * b
        case "division" | "d" if b != 0:
            return a / b
        case "division" | "d":
            return "Error: división por cero"
        case _:
            return "Operación no soportada"

print("suma 10 3 →", calcular("suma", 10, 3))
print("d 10 0 →", calcular("d", 10, 0))
print("potencia →", calcular("potencia", 2, 3))

# Clasificación de un punto (patrones de tupla)
def clasificar_punto(punto):
    match punto:
        case (0, 0):
            return "origen"
        case (0, y):
            return f"eje Y en {y}"
        case (x, 0):
            return f"eje X en {x}"
        case (x, y) if x == y:
            return "diagonal"
        case (x, y):
            return f"punto ({x}, {y})"
        case _:
            return "no es un punto 2D"

print((0, 0), "→", clasificar_punto((0, 0)))
print((0, 5), "→", clasificar_punto((0, 5)))
print((3, 3), "→", clasificar_punto((3, 3)))
print((1, 2), "→", clasificar_punto((1, 2)))

# Lista de longitud variable
def resumir(secuencia):
    match secuencia:
        case []:
            return "vacía"
        case [unico]:
            return f"un elemento: {unico}"
        case [primero, *resto]:
            return f"empieza por {primero}, resto={resto}"

print(resumir([]), "|", resumir([7]), "|", resumir([1, 2, 3]))
````

## Epígrafe 3 — `while`

### 3.1 Conceptos clave

| Concepto | Definición |
|---|---|
| **`while`** | Repite un bloque **mientras** la condición sea verdadera. Se evalúa **antes** de cada iteración. |
| **Contador** | Variable que cuenta iteraciones. Hay que **actualizarla** o el bucle no termina. |
| **Acumulador** | Variable que va guardando una suma, un producto, un texto, etc. |
| **`break`** | Sale **inmediatamente** del bucle más interno. |
| **`continue`** | Salta el resto del cuerpo y vuelve a evaluar la condición. |
| **`while True`** | Bucle «infinito» controlado: se sale con `break` (menús, validación de entrada). |
| **Bucle infinito** | La condición nunca se vuelve falsa (olvidar el contador es el clásico). |

```python
while condicion:
    # cuerpo
    # actualización
```


### 3.2 Explicación razonada

#### Qué ocurre

1. Se evalúa la condición.
2. Si es falsa, el cuerpo **no se ejecuta ni una vez** (es un bucle pre-test).
3. Si es verdadera, se ejecuta el cuerpo y se vuelve al paso 1.

Por eso `while` responde a «¿sigo?», no a «¿cuántos elementos hay?». Si el número de vueltas depende de un iterable conocido, `for` es más claro. Si depende de una condición que cambia (`seguir`, `opcion != "salir"`, `intento < 3`), `while` es la herramienta.

#### Contador y actualización

```python
contador = 0
while contador < 5:
    print(contador)
    contador += 1
```

Sin `contador += 1` la condición `contador < 5` es **siempre** verdadera → bucle infinito. Ese error no es de sintaxis: el programa «funciona» y no termina.

#### `while True` y `break`

Un menú o una validación no saben a priori cuántas veces van a repetir. El patrón idiomático es:

```python
while True:
    opcion = pedir()
    if opcion == "salir":
        break
    procesar(opcion)
```

`while True` **no** es sucio si hay una salida clara. Es sucio si la salida está escondida o no existe.

#### `continue`

Sirve para «saltar este caso y seguir». Ejemplo: acumular solo positivos. Abusar de `continue` (y de `break`) convierte el cuerpo en un laberinto; a veces un `if` anidado es más legible.

#### Validación de entrada (idea)

En un script real usarías `input()`. En este cuaderno **no** llamamos a `input()` en celdas automáticas (bloquearía la ejecución). El patrón es el mismo:

```python
while True:
    texto = input("Número: ")
    if texto.isdigit():
        n = int(texto)
        break
    print("No válido")
```

#### Cuándo NO usar `while`

- Recorrer `lista`, `str`, `dict`, `range` → `for`.
- «Repetir 10 veces» con un contador que solo imita a `range` → `for i in range(10)`.
- Recursión donde un bucle basta.

#### `else` en `while`

Python permite `while ... else`. El `else` se ejecuta si el bucle **terminó sin `break`**. Es útil en búsquedas; mucha gente lo desconoce y por eso aparece en examen.


````python
# ── Práctica guiada 3 — while ────────────────────────────────────────────────
# 1) Contador ascendente
c = 0
asc = []
while c < 4:
    asc.append(c)
    c += 1
print("ascendente:", asc)

# 2) Contador descendente
c = 3
desc = []
while c > 0:
    desc.append(c)
    c -= 1
print("descendente:", desc)

# 3) Acumulador: suma 1+2+...+5
n = 1
total = 0
while n <= 5:
    total += n
    n += 1
print("suma 1..5 =", total)

# 4) break: parar al encontrar un negativo
valores = [3, 1, -2, 8]
i = 0
while i < len(valores):
    if valores[i] < 0:
        print("negativo en índice", i)
        break
    i += 1
else:
    print("no había negativos")  # no se ejecuta: hubo break

# 5) continue: acumular solo pares
nums = [1, 2, 3, 4, 5, 6]
i = 0
pares = 0
while i < len(nums):
    actual = nums[i]
    i += 1
    if actual % 2 != 0:
        continue
    pares += actual
print("suma de pares =", pares)

# 6) Simulación de menú (sin input): se sale con break
opciones = ["ver", "ver", "salir", "ver"]  # cola simulada
idx = 0
historial = []
while True:
    op = opciones[idx]
    idx += 1
    if op == "salir":
        break
    historial.append(op)
print("historial menú:", historial)

# 7) Validación: aceptar el primer entero en rango
entradas = ["hola", "99", "7", "3"]
idx = 0
elegido = None
while True:
    bruto = entradas[idx]
    idx += 1
    if bruto.isdigit():
        n = int(bruto)
        if 1 <= n <= 10:
            elegido = n
            break
print("entero válido:", elegido)

print("\n--- El siguiente ejemplo NO se ejecuta: es un bucle infinito ---")
print("contador = 0")
print("while contador < 5:")
print("    print(contador)")
print("# falta contador += 1  →  la condición nunca se vuelve falsa")
````

| **`enumerate(iterable)`** | Pares `(índice, elemento)` empezando en 0. |
| **`dict.keys()`** | Vista de las **claves**. |
| **`dict.values()`** | Vista de los **valores**. |
| **`dict.items()`** | Vista de las **parejas** `(clave, valor)`. |
| **`break` / `continue`** | Igual que en `while`, aplicados al `for` más interno. |

```text
dict
 |
 +-- keys()   →  claves
 +-- values() →  valores
 +-- items()  →  parejas (clave, valor)
```

```python
for clave in alumno.keys():      # solo nombres de campo
    ...
for valor in alumno.values():    # solo contenidos
    ...
for clave, valor in alumno.items():  # los dos
    ...
```


### 4.2 Explicación razonada

#### Qué ocurre

`for elemento in iterable:` pide un iterador al objeto y va llamando al siguiente elemento hasta que se acaba. No necesitas índice. Por eso `for letra in "Ana"` recorre caracteres, `for n in [8, 7, 9]` recorre notas y `for clave in diccionario` recorre **claves** (sí: iterar un dict sin `.keys()` recorre claves).

#### Por qué `for` y no `while`

El número de vueltas lo dicta el iterable. No hay que actualizar un contador. El error clásico de `while` (olvidar `i += 1`) **desaparece**.

#### `range`

`range` no materializa una lista enorme: es perezoso. `range(3)` produce 0, 1, 2. `range(2, 5)` produce 2, 3, 4. `range(10, 0, -2)` produce 10, 8, 6, 4, 2.

Úsalo cuando necesites **el índice** o un recuento. Si solo necesitas los elementos, recorre la lista directamente.

#### `enumerate`

```python
for i, nota in enumerate(notas):
    print(i, nota)
```

Evita `for i in range(len(notas)): notas[i]`, más ruidoso y más fácil de equivocar.

#### Diccionarios: el error de examen más frecuente

Un alumno es un diccionario. Un grupo es una **lista de diccionarios**. Hay que saber qué recorres:

```python
alumno = {"nombre": "Ana", "nota": 8}

for clave in alumno:            # "nombre", "nota"
for clave in alumno.keys():     # igual, explícito
for valor in alumno.values():   # "Ana", 8
for k, v in alumno.items():     # ("nombre", "Ana"), ("nota", 8)
```

Si escribes `for nombre, nota in alumno.values()` **no** obtienes nombre y nota: `values()` da un valor por vuelta, no una pareja. El desempaquetado fallará o hará algo extraño.

Si escribes `for k, v in alumno` intentas desempaquetar **cada clave** (un string) en dos variables → error típico.

#### Lista de diccionarios (dataset)

```python
for alumno in alumnos:          # cada alumno es un dict
    print(alumno["nombre"], alumno["notas"])
```

Aquí el `for` recorre la **lista**. Dentro, accedes por clave. Mezclar `.items()` del alumno con el `for` del grupo es el lío habitual.

#### `break` / `continue` / `else`

Igual que en `while`. El `else` del `for` se ejecuta si no hubo `break` (búsqueda infructuosa).

#### Cuándo NO

- Esperar a una condición externa (`input`, socket) → `while`.
- Recorrer y **borrar** elementos de la misma lista que recorres: crea una copia o construye otra lista.


````python
# ── Práctica guiada 4 — for ──────────────────────────────────────────────────
notas = [8, 7, 9, 4]
print("lista:", end=" ")
for n in notas:
    print(n, end=" ")
print()

print("tupla:", tuple(x for x in (1, 2, 3)))
print("string:", [c for c in "Ana"])
print("range(3):", list(range(3)))
print("range(2, 8, 2):", list(range(2, 8, 2)))

print("enumerate:")
for i, n in enumerate(notas):
    print(f"  índice {i} → {n}")

print("media =", sum(notas) / len(notas))
print("máximo =", max(notas), "mínimo =", min(notas))

aprobados = 0
for n in notas:
    if n >= 5:
        aprobados += 1
print("aprobados:", aprobados)

# Diccionario de un alumno
alumno = {"nombre": "Ana", "edad": 20, "nota": 8}
print("\n--- keys ---")
for clave in alumno.keys():
    print(" clave:", clave)

print("--- values ---")
for valor in alumno.values():
    print(" valor:", valor)

print("--- items ---")
for clave, valor in alumno.items():
    print(f" {clave} = {valor}")

print("for alumno (sin método) recorre claves:", list(alumno))

# Dataset: lista de diccionarios
alumnos = [
    {"nombre": "Ana", "notas": [8, 7, 9]},
    {"nombre": "Luis", "notas": [4, 5, 3]},
    {"nombre": "Pedro", "notas": [9, 10, 8]},
]
print("\n--- dataset ---")
for persona in alumnos:
    media = sum(persona["notas"]) / len(persona["notas"])
    print(f" {persona['nombre']}: media={media:.2f}")

# break / continue
print("primera nota < 5:", end=" ")
for n in notas:
    if n < 5:
        print(n)
        break
else:
    print("todas >= 5")
````

# PARTE II — FUNCIONES

Una función empaqueta un algoritmo con un **nombre**, unos **parámetros** y, casi siempre, un **valor de retorno**. El resto del curso (módulos, APIs, tests) se sostiene sobre esta idea.

Cuatro verbos que no debes mezclar:

| Verbo | Código | Qué ocurre |
|---|---|---|
| **Definir** | `def sumar(a, b):` | Se crea el objeto función. **No** se ejecuta el cuerpo. |
| **Llamar** | `sumar(5, 3)` | Se ejecuta el cuerpo con esos argumentos. |
| **Devolver** | `return a + b` | La llamada **produce un valor**. |
| **Guardar** | `resultado = sumar(5, 3)` | Ese valor queda en un nombre. |


---
## Epígrafe 5 — `def`, llamada y `return`

### 5.1 Conceptos clave

| Concepto | Definición |
|---|---|
| **`def`** | Crea una función y la ata a un nombre. |
| **Parámetro** | Nombre en la definición: `def sumar(a, b)`. |
| **Argumento** | Valor en la llamada: `sumar(5, 3)`. |
| **`return`** | Termina la función y entrega un valor al llamador. |
| **`None`** | Lo que «devuelve» una función si no hay `return` (o hay `return` sin valor). |
| **Docstring** | Cadena en la primera línea del cuerpo: documenta qué hace, no cómo. |
| **`print` vs `return`** | `print` muestra en pantalla. `return` comunica un valor al resto del programa. |

```python
def sumar(a, b):
    '''Devuelve la suma de a y b.'''
    return a + b

resultado = sumar(5, 3)   # resultado vale 8
```


### 5.2 Explicación razonada

#### Definir no es llamar

Al ejecutar `def sumar(a, b):`, Python crea un **objeto función** y lo guarda en el nombre `sumar`. El cuerpo no corre. Por eso puedes definir funciones en cualquier orden relativo a su uso posterior (mientras definas **antes** de llamar).

#### Llamar

`sumar(5, 3)`:

1. Crea un marco local.
2. Ata `a → 5`, `b → 3`.
3. Ejecuta el cuerpo.
4. El `return` destruye el marco y **sustituye** la llamada por el valor.

`resultado = sumar(5, 3)` guarda **8**, no la función. `operacion = sumar` (sin paréntesis) guarda la función. Esa diferencia vuelve en los epígrafes 15-17 y cae en todos los exámenes.

#### `print` no es `return`

```python
def malo(a, b):
    print(a + b)

x = malo(5, 3)
```

En pantalla ves `8`. `x` vale **`None`**. Si luego haces `x + 1`, hay `TypeError`. En un notebook el `print` engaña: «parece que funciona».

Una función de cálculo debe **devolver**. El `print` es de presentación, no de interfaz.

#### Función sin `return`

```python
def saludar(nombre):
    print(f"Hola, {nombre}")
```

Es válida: hace un efecto. Su valor es `None`. No la uses como si devolviera el saludo.

#### Varios `return`

`return` **sale**. Las líneas posteriores del mismo camino no se ejecutan. Varios `return` (uno por rama) son claros en clasificaciones; no hace falta un solo `return` al final por dogma.

#### Docstring y modularidad

La docstring (la cadena inicial del cuerpo, entre comillas triples) queda en `funcion.__doc__`. Documenta **contrato**: qué recibe, qué devuelve, qué no hace. La reutilización nace de funciones cortas con un propósito: `media(notas)`, `aprobado(media)`, `analizar(alumno)` — no de un `main` de 200 líneas.

#### Cuándo definir una función

- El mismo cálculo aparece dos veces.
- Quieres testear un fragmento.
- El nombre **explica** mejor que un comentario.

#### Cuándo NO

- Una sola expresión usada una vez: innecesario.
- Función que solo llama a otra sin aportar nombre útil: ruido.


````python
# ── Práctica guiada 5 — def / llamada / return ───────────────────────────────
def sumar(a, b):
    """Devuelve la suma de a y b."""
    return a + b

print("definición: sumar =", sumar)
print("llamada: sumar(5, 3) =", sumar(5, 3))
resultado = sumar(5, 3)
print("guardado:", resultado, "tipo:", type(resultado).__name__)

def solo_print(a, b):
    print("  print interno:", a + b)

x = solo_print(5, 3)
print("lo que quedó en x:", x)

def sin_return(a, b):
    a + b

print("sin return →", sin_return(5, 3))

def clasificar(n):
    if n < 0:
        return "negativo"
    if n == 0:
        return "cero"
    return "positivo"

print(clasificar(-1), clasificar(0), clasificar(4))
print("docstring:", sumar.__doc__)
````

## Epígrafe 6 — Parámetros y argumentos

### 6.1 Conceptos clave

**Parámetro**: hueco en la definición.

```python
def presentar(nombre, edad, ciudad):
    ...
```

**Argumento**: valor concreto en la llamada.

```python
presentar("Ana", 20, "Valladolid")
```

| Tipo de argumento | Ejemplo | Cómo se ata |
|---|---|---|
| **Posicional** | `presentar("Ana", 20, "Valladolid")` | Por **orden**. |
| **Por nombre** (keyword) | `presentar(ciudad="Valladolid", nombre="Ana", edad=20)` | Por **nombre** del parámetro. |
| **Mezcla** | `presentar("Ana", ciudad="Valladolid", edad=20)` | Primero todos los posicionales; después los nombrados. |

Regla de oro: **los posicionales van antes que los nombrados**. No puedes hacer `presentar(nombre="Ana", 20, "Valladolid")`.


### 6.2 Explicación razonada

Al llamar, Python construye un mapa parámetro → objeto.

- Posicional: el 1.º argumento entra en el 1.º parámetro, etc. El **orden es el contrato**.
- Por nombre: el orden **deja de importar**; el nombre **es** el contrato. Ideal cuando hay muchos parámetros o valores del mismo tipo (`int`, `int`) que se confunden.
- Mezcla: los posicionales ocupan los primeros huecos; los nombrados cubren el resto. Un parámetro no puede recibir **dos** valores (posicional y nombre) → `TypeError`. Un parámetro sin valor y sin defecto → `TypeError: missing argument`.

```python
def presentar(nombre, edad, ciudad):
    return f"{nombre}, {edad}, {ciudad}"

presentar("Ana", 20, "Valladolid")
presentar(ciudad="Valladolid", nombre="Ana", edad=20)
presentar("Ana", ciudad="Valladolid", edad=20)   # válido
# presentar("Ana", nombre="Luis", edad=20)       # nombre duplicado
# presentar(edad=20, "Ana", ciudad="V")          # posicional después de keyword: SyntaxError
```

#### Cuándo posicional

Funciones de 1-2 argumentos con orden obvio: `max(a, b)`, `saludar("Ana")`.

#### Cuándo por nombre

Tres o más, o booleanos (`reverse=True`), o cuando lees la llamada y no se entiende el 2.º `True`.

#### Cuándo NO mezclar a ciegas

Si ya nombraste un parámetro, no pases otro posicional **después**. El parser no lo permite.


````python
# ── Práctica guiada 6 — parámetros vs argumentos ─────────────────────────────
def presentar(nombre, edad, ciudad):
    return f"{nombre} tiene {edad} años y vive en {ciudad}"

print("posicional:", presentar("Ana", 20, "Valladolid"))
print("por nombre:", presentar(ciudad="Valladolid", nombre="Ana", edad=20))
print("mezcla:", presentar("Ana", ciudad="Valladolid", edad=20))

def probar(llamada):
    try:
        print(llamada(), "← ok")
    except Exception as e:
        print(type(e).__name__ + ":", e)

probar(lambda: presentar("Ana", nombre="Luis", edad=20, ciudad="Valladolid"))
probar(lambda: presentar("Ana", 20))
probar(lambda: presentar())
````

## Epígrafe 7 — Argumentos por defecto

### 7.1 Conceptos clave

```python
def saludar(nombre="Usuario"):
    return f"Hola, {nombre}"
```

| Llamada | `nombre` vale |
|---|---|
| `saludar()` | `"Usuario"` (defecto) |
| `saludar("Ana")` | `"Ana"` |
| `saludar(nombre="Luis")` | `"Luis"` |

El valor por defecto se usa **solo** si ese parámetro no recibe argumento.

Los parámetros con defecto van **después** de los que no lo tienen: `def f(a, b=1)` sí; `def f(a=1, b)` no.


### 7.2 Explicación razonada

El defecto se evalúa **una sola vez**, cuando se ejecuta el `def`, no en cada llamada. Eso es irrelevante para `None`, `0` o `"Usuario"` (inmutables). Es una trampa con **listas, diccionarios y sets**.

#### El problema del mutable por defecto

```python
def malo(lista=[]):
    lista.append(1)
    return lista

print(malo())  # [1]
print(malo())  # [1, 1]  ← la misma lista
```

`[]` del `def` es **un** objeto, guardado en `malo.__defaults__`. Cada llamada sin argumento **reutiliza** esa lista y hace `append`. No es una lista nueva.

#### La solución idiomática

```python
def bueno(lista=None):
    if lista is None:
        lista = []
    lista.append(1)
    return lista
```

`None` es inmutable (hay un solo `None`, pero no se modifica). **Dentro** de la llamada creas una lista nueva.

Atención: si el llamador pasa una lista, `bueno` **sí** la muta (hace `append`). El defecto solo arregla el caso «no me pasaron lista». Si no quieres mutar el argumento, copia: `lista = list(lista)`.

#### Cuándo usar defectos

Valores opcionales reales (`modo="auto"`, `sep=" "`). No para «olvidé el argumento» de un dato obligatorio.


````python
# ── Práctica guiada 7 — valores por defecto ──────────────────────────────────
def saludar(nombre="Usuario"):
    return f"Hola, {nombre}"

print(saludar())
print(saludar("Ana"))

def malo(lista=[]):
    lista.append(1)
    return lista

print("malo 1.ª llamada:", malo())
print("malo 2.ª llamada:", malo())
print("malo 3.ª con lista nueva:", malo([]))
print("malo 4.ª sin args (sigue la compartida):", malo())
print("defaults internas:", malo.__defaults__)

def bueno(lista=None):
    if lista is None:
        lista = []
    lista.append(1)
    return lista

print("bueno 1:", bueno())
print("bueno 2:", bueno())
externa = [10]
print("bueno muta la pasada:", bueno(externa), "externa=", externa)
````

## Epígrafe 8 — Argumentos por nombre

### 8.1 Conceptos clave

Llamada **posicional**:

```python
funcion("Ana", 20)
```

Llamada **por nombre** (keyword arguments):

```python
funcion(nombre="Ana", edad=20)
```

| | Posicional | Por nombre |
|---|---|---|
| Contrato | El **orden** | El **nombre** |
| Legibilidad | Alta si hay 1-2 args obvios | Alta si hay muchos o booleanos |
| Riesgo | Invertir dos `int` | Typo en el nombre (`nambre=`) |
| Reordenación | Cambia el significado | Irrelevante |


### 8.2 Explicación razonada

En la práctica profesional se mezclan: posicionales para lo principal, nombres para lo opcional.

```python
sorted(alumnos, key=lambda a: a["nota"], reverse=True)
```

`key` y `reverse` **deben** ir por nombre: si `sorted` recibiera dos posicionales extra, no sabría qué es cada uno. De hecho, tras `*`, muchos parámetros son *keyword-only*.

Ventajas del nombre:

- La llamada se documenta sola: `conectar(host="localhost", puerto=5432)`.
- Añadir un parámetro con defecto en medio **no** rompe llamadas por nombre.
- Evitas `f(True, False, True)`.

#### Keyword-only (avance breve)

```python
def f(a, *, estricto=False):
    ...
f(1, estricto=True)   # ok
f(1, True)            # TypeError
```

`*` marca: lo que sigue **solo** por nombre. Útil para banderas.

#### Cuándo NO

`sumar(a=5, b=3)` es correcto pero ruidoso. El estilo PEP 8: posicional si el orden es parte obvia del significado (`min(a, b)`).


````python
# ── Práctica guiada 8 — llamadas por nombre ──────────────────────────────────
def funcion(nombre, edad):
    return f"{nombre}:{edad}"

print("posicional:", funcion("Ana", 20))
print("por nombre:", funcion(nombre="Ana", edad=20))
print("orden invertido:", funcion(edad=20, nombre="Ana"))

def f(a, *, estricto=False):
    return a, estricto

print("keyword-only ok:", f(1, estricto=True))
try:
    f(1, True)
except TypeError as e:
    print("keyword-only posicional:", e)
````

### 8.4 Errores frecuentes

**1. Posicional tras nombre** — `SyntaxError`.

**2. Nombre que no existe** — `TypeError: unexpected keyword argument`.

**3. Duplicar** — posicional + el mismo keyword.

**4. Creer que el nombre «busca» en el llamador**

```python
nombre = "Ana"
funcion(nombre)   # ok, posicional
funcion(nombre=nombre)  # el de la izquierda es el PARÁMETRO
```

El de la izquierda del `=` es el parámetro; el de la derecha es el valor (aquí, la variable del llamador).

## Epígrafe 9 — Paso de argumentos y referencias

### 9.1 Conceptos clave

Python **no** se explica bien con la dicotomía «paso por valor / paso por referencia» de C/Java.

Modelo preciso: **paso de objetos por referencia compartida** (*pass-by-object-reference* / *pass-by-sharing*).

| Idea | Significado |
|---|---|
| **Objeto** | Dato en memoria (`10`, `[1, 2]`, `"Ana"`). |
| **Nombre** | Etiqueta que apunta a un objeto (`x`, `numeros`). |
| **Referencia** | Esa flecha nombre → objeto. |
| **Mutable** | El objeto puede cambiar por dentro (`list`, `dict`, `set`). |
| **Inmutable** | Cualquier «cambio» produce **otro** objeto (`int`, `str`, `tuple`). |

Al llamar, el parámetro se convierte en **otro nombre** para **el mismo objeto** que el argumento.


### 9.2 Explicación razonada

#### Caso 1 — inmutable + reasignación del parámetro

```python
def cambiar(numero):
    numero = 100

x = 10
cambiar(x)
print(x)   # 10
```

`numero` y `x` apuntan al `10`. `numero = 100` hace que **solo `numero`** apunte a `100`. `x` sigue en `10`. No hay forma de «cambiar el 10»: los `int` son inmutables.

#### Caso 2 — mutable + mutación in-place

```python
def cambiar(lista):
    lista.append(100)

numeros = [1, 2, 3]
cambiar(numeros)
print(numeros)   # [1, 2, 3, 100]
```

`lista` y `numeros` son dos nombres del **mismo** objeto lista. `append` no crea otra lista: modifica esa. El llamador **sí** ve el cambio.

#### Caso 3 — mutable + reasignación del parámetro

```python
def cambiar(lista):
    lista = [100, 200, 300]

numeros = [1, 2, 3]
cambiar(numeros)
print(numeros)   # [1, 2, 3]
```

`lista = [...]` pega la etiqueta `lista` a **otro** objeto. `numeros` no se enteró. Mutar ≠ reasignar.

```text
Antes:     numeros ──┐
                     ├──► [1, 2, 3]
           lista   ──┘

Tras lista.append(100): ambos siguen apuntando a [1, 2, 3, 100]

Tras lista = [100, 200, 300]:
           numeros ──► [1, 2, 3]
           lista   ──► [100, 200, 300]
```

#### Cuándo mutar dentro de una función

- Acumuladores explícitos (`registrar(evento, historial)`).
- APIs que documentan «modifica in-place» (`list.sort()`).

#### Cuándo NO

- Si el llamador no espera el efecto (`media` no debe vaciar la lista).
- Prefiere devolver un objeto nuevo: más fácil de razonar y de testear.

`+=` en listas muta (`extend` interno); en enteros reasigna. Otro clásico de examen.


````python
# ── Práctica guiada 9 — tres casos de referencias ────────────────────────────
def cambiar_numero(numero):
    numero = 100

x = 10
cambiar_numero(x)
print("caso int, reasignar parámetro → x =", x)

def cambiar_lista_append(lista):
    lista.append(100)

numeros = [1, 2, 3]
cambiar_lista_append(numeros)
print("caso list, append → numeros =", numeros)

def cambiar_lista_reasignar(lista):
    lista = [100, 200, 300]

numeros2 = [1, 2, 3]
cambiar_lista_reasignar(numeros2)
print("caso list, reasignar → numeros2 =", numeros2)

def tocar_id(lista):
    print("  id dentro, antes =", id(lista))
    lista.append(0)
    print("  id dentro, después (append) =", id(lista))
    lista = [9]
    print("  id dentro, tras reasignar =", id(lista))

externa = [1]
print("id fuera, antes =", id(externa))
tocar_id(externa)
print("id fuera, después =", id(externa), "contenido =", externa)
````

## Epígrafe 10 — Funciones dentro de funciones

### 10.1 Conceptos clave

```python
def exterior():
    def interior():
        ...
    interior()
```

| Idea | Detalle |
|---|---|
| **Función interna** | Se define **dentro** del cuerpo de otra. |
| **Ámbito** | `interior` no existe como nombre global; nace en cada llamada a `exterior`. |
| **Cierre (closure)** | La interna puede **usar** variables de la externa. |
| **Utilidad** | Helpers privados, elegir una operación, fabricar funciones (epígrafes 16-17). |


### 10.2 Explicación razonada

Cada vez que llamas a `exterior()`, se **vuelve a definir** `interior` (otro objeto función). No es un coste que deba obsesionarte; sí explica que dos llamadas no comparten «la misma» interna salvo que la **devuelvas** y la guardes.

La interna ve las variables de la externa (ámbito *enclosing*). Eso permite:

```python
def operar(orden):
    def incrementar(n):
        return n + 1
    def decrementar(n):
        return n - 1
    match orden:
        case "i":
            return incrementar
        case "d":
            return decrementar
```

`return incrementar` entrega **la función**, no un número. `return incrementar(1)` entregaría `2`. El llamador hace:

```python
var_funcion = operar("i")
print(var_funcion(6))   # 7
```

Paso a paso:

1. `operar("i")` define dos internas y **devuelve** `incrementar`.
2. `var_funcion` apunta a esa función.
3. `var_funcion(6)` es `incrementar(6)`.

#### Utilidad

- Ocultar helpers que no merecen ser públicos.
- Fábricas: «dame la operación y te doy la función lista».
- Closures con estado (epígrafe 11, `nonlocal`).

#### Cuándo NO

Anidar cinco niveles. Si `interior` sirve en otro sitio, sácala al módulo.


````python
# ── Práctica guiada 10 — funciones anidadas ──────────────────────────────────
def operar(orden):
    def incrementar(n):
        return n + 1

    def decrementar(n):
        return n - 1

    match orden:
        case "i":
            return incrementar
        case "d":
            return decrementar
        case _:
            return None

var_funcion = operar("i")
print("var_funcion es callable:", callable(var_funcion))
print("var_funcion(6) =", var_funcion(6))
print("operar('d')(6) =", operar("d")(6))

def exterior():
    mensaje = "desde fuera"

    def interior():
        return mensaje

    return interior()

print("acceso enclosing:", exterior())

try:
    incrementar(1)
except NameError as e:
    print("incrementar no es global:", e)
````

**1. `return incrementar(1)` cuando querías devolver la función**

Devuelves un `int`. Luego `var_funcion(6)` → `TypeError: 'int' object is not callable`.

**2. `return incrementar()` sin argumentos**

Si `incrementar` espera `n`, `TypeError`.

**3. Llamar a la interna desde fuera**

`interior` no está en el global. `NameError`.

**4. Olvidar llamar a la interna *dentro* si era un helper**


## Epígrafe 11 — Ámbito de variables (LEGB)

### 11.1 Conceptos clave

Cuando Python ve un nombre, busca en este orden:

| Letra | Ámbito | Dónde |
|---|---|---|
| **L** | *Local* | Función actual. |
| **E** | *Enclosing* | Funciones envolventes (internas). |
| **G** | *Global* | Módulo. |
| **B** | *Built-in* | `len`, `print`, `range`, `None`... |

| Palabra | Efecto |
|---|---|
| **`global x`** | Las asignaciones a `x` en esta función van al **módulo**. |
| **`nonlocal x`** | Las asignaciones van al `x` de una **función envolvente**, no al global. |

Sin `global`/`nonlocal`, **asignar** `x = ...` crea un `x` **local** (toda la función).


### 11.2 Explicación razonada

#### Lectura vs escritura

**Leer** una variable externa está permitido:

```python
x = "global"

def f():
    print(x)   # lee la global
```

**Asignar** sin declaración crea un local. Python decide eso **en tiempo de compilación** de la función: si hay un `x =` en cualquier parte del cuerpo, `x` es local **en toda** la función.

```python
x = 10

def f():
    print(x)  # UnboundLocalError: x se asigna más abajo, luego es local
    x = 20
```

Ese `UnboundLocalError` no es caprichoso: al hacer `print(x)`, el `x` local **aún no tiene valor**.

#### El ejemplo LEGB clásico

```python
x = "global"

def exterior():
    x = "enclosing"

    def interior():
        x = "local"
        print(x)

    interior()

exterior()   # imprime "local"
```

Cada `x =` crea un nombre **en su función**. No pisan el de fuera salvo `global`/`nonlocal`.

#### `global`

```python
contador = 0

def tick():
    global contador
    contador += 1
```

Sin `global`, `contador += 1` es «leer y asignar un local» → `UnboundLocalError`.

`global` acopla la función al módulo. Úsalo poco: mejor devolver valores o un closure.

#### `nonlocal` y closures

```python
def crear_contador():
    cuenta = 0

    def siguiente():
        nonlocal cuenta
        cuenta += 1
        return cuenta

    return siguiente

contador = crear_contador()
print(contador())  # 1
print(contador())  # 2
print(contador())  # 3
```

`cuenta` vive en `crear_contador`. `siguiente` la **modifica** gracias a `nonlocal`. Cada `crear_contador()` fabrica su **propio** `cuenta`: dos contadores son independientes.

`return siguiente` (la función, no `siguiente()`).

#### Built-ins

Si haces `list = 3`, rompes `list()` en ese ámbito. No sombrees `str`, `len`, `id`, `type`.

#### Cuándo NO

- `global` para «compartir todo»: convierte el programa en estado spaghetti.
- `nonlocal` en funciones que no están anidadas: `SyntaxError`.


````python
# ── Práctica guiada 11 — LEGB, global, nonlocal ──────────────────────────────
x = "global"

def exterior():
    x = "enclosing"

    def interior():
        x = "local"
        print("interior:", x)

    interior()
    print("exterior:", x)

exterior()
print("módulo:", x)

# UnboundLocalError: x es local en TODA la función porque más abajo se asigna
def trampa():
    print(x)
    x = "local"

try:
    trampa()
except UnboundLocalError as e:
    print("UnboundLocalError:", e)

contador = 0

def tick():
    global contador
    contador += 1
    return contador

print("tick", tick(), tick())

def crear_contador():
    cuenta = 0

    def siguiente():
        nonlocal cuenta
        cuenta += 1
        return cuenta

    return siguiente

a = crear_contador()
b = crear_contador()
print("a:", a(), a(), "b:", b(), "a otra vez:", a())
````

## Epígrafe 12 — `*args`

### 12.1 Conceptos clave

```python
def sumar(*numeros):
    ...
```

El `*` en la **definición** empaqueta los argumentos **posicionales sobrantes** en una **tupla**.

| Llamada | `numeros` / `args` |
|---|---|
| `mostrar()` | `()` |
| `mostrar(10)` | `(10,)` |
| `mostrar(10, 20, 30)` | `(10, 20, 30)` |

El nombre `args` es convención; lo que importa es el `*`.


### 12.2 Explicación razonada

`*args` responde a «no sé cuántos posicionales me van a pasar». `sumar(1, 2)` y `sumar(1, 2, 3, 4)` usan la misma definición.

Dentro, `args` es una **tupla**: inmutable, ordenada, recorrible con `for`.

```python
def mostrar(*args):
    print(args)

mostrar(10, 20, 30)   # (10, 20, 30)
```

También puedes combinar un posicional fijo con el resto:

```python
def media_desde(etiqueta, *valores):
    ...
media_desde("examen", 8, 7, 9)
# etiqueta="examen", valores=(8, 7, 9)
```

#### Unpacking en la llamada

El `*` también **desempaqueta** al llamar:

```python
def f(a, b, c):
    return a + b + c
xs = [1, 2, 3]
f(*xs)   # equivalente a f(1, 2, 3)
```

No confundas los dos usos: en `def` agrupa; en la llamada, expande.

#### Cuándo usarlo

Suma, máximo, `print`-likes, wrappers. Estadística de un número variable de notas.

#### Cuándo NO

Si siempre son exactamente 2 argumentos, pon `a, b`. `*args` borra el contrato. Tampoco uses `*args` para pasar un dict de alumno: eso es `**kwargs`.


````python
# ── Práctica guiada 12 — *args ───────────────────────────────────────────────
def mostrar(*args):
    print("args =", args, "tipo =", type(args).__name__)

mostrar(10, 20, 30)
mostrar()
mostrar("Ana")

def estadistica(*numeros):
    if not numeros:
        return None
    return {
        "n": len(numeros),
        "suma": sum(numeros),
        "media": sum(numeros) / len(numeros),
        "maximo": max(numeros),
        "minimo": min(numeros),
    }

print(estadistica(8, 7, 9, 4))

def f(a, b, c):
    return a, b, c

print("unpacking de lista:", f(*[1, 2, 3]))
````

### 12.4 Errores frecuentes

**1. Confundir `*args` con `**kwargs`**

`mostrar(nombre="Ana")` no entra en `*args` → `TypeError: unexpected keyword`.

**2. Tratar `args` como lista mutable**

Es tupla: `args.append(1)` → `AttributeError`. Usa `list(args)` si necesitas mutar una copia.

**3. Olvidar la coma de la tupla de un elemento**

`args` con un argumento es `(10,)`, no `10`. `len(args)` es 1.

**4. `def f(*args, a):` y llamar `f(1, 2)`**

`a` queda keyword-only. `f(1, 2)` mete todo en `args` y falta `a`.

## Epígrafe 13 — `**kwargs`

### 13.1 Conceptos clave

```python
def crear_alumno(**datos):
    return datos
```

`**` en la definición empaqueta los argumentos **por nombre** extra en un **diccionario**.

```python
alumno = crear_alumno(
    nombre="Ana",
    edad=20,
    nota=8,
    ciudad="Valladolid",
)
# datos == {"nombre": "Ana", "edad": 20, "nota": 8, "ciudad": "Valladolid"}
```

| Recorrido | Qué obtienes |
|---|---|
| `for clave in datos.keys():` | solo claves (`"nombre"`, `"edad"`, …) |
| `for valor in datos.values():` | solo valores (`"Ana"`, `20`, …) |
| `for clave, valor in datos.items():` | parejas |

`for clave in datos:` equivale a `.keys()`.


### 13.2 Explicación razonada

`**kwargs` es el espejo de `*args` para nombres. Construye fichas, opciones, wrappers de APIs.

Las claves son **strings** (los nombres de los argumentos). Los valores, lo que hayas pasado.

Al llamar, `**` **expande** un dict:

```python
def f(nombre, edad):
    ...
d = {"nombre": "Ana", "edad": 20}
f(**d)
```

Si el dict tiene una clave que no es parámetro y no hay `**kwargs`, `TypeError`.

#### Cuándo

Campos opcionales de un alumno, configuración (`debug=True, timeout=5`), reenviar opciones a otra función.

#### Cuándo NO

Tres campos fijos y obligatorios: pon parámetros normales. Un dict anónimo pierde el chequeo de nombres erróneos… salvo que lo valides tú.


````python
# ── Práctica guiada 13 — **kwargs ────────────────────────────────────────────
def crear_alumno(**datos):
    return datos

alumno = crear_alumno(nombre="Ana", edad=20, nota=8, ciudad="Valladolid")
print("alumno:", alumno)
print("claves:", list(alumno.keys()))
print("valores:", list(alumno.values()))
print("items:")
for clave, valor in alumno.items():
    print(f"  {clave} = {valor}")

def ficha(nombre, **extra):
    return nombre, extra

print("fijo + kwargs:", ficha("Luis", edad=21, nota=5))
````

### 13.4 Errores frecuentes

**1. Pasar posicionales a una función que solo tiene `**kwargs`**

`crear_alumno("Ana")` → `TypeError: takes 0 positional arguments`.

**2. Recorrer `.values()` esperando pares**

Ya visto con dicts: usa `.items()`.

**3. Confundir `kwargs` con `args`**

`kwargs` no es una tupla. `kwargs[0]` no es el primer argumento nombrado (las claves no están indexadas así).

**4. Clave repetida al expandir**

`f(nombre="Ana", **{"nombre": "Luis"})` → `TypeError` duplicado.

## Epígrafe 14 — `*args` + `**kwargs`

### 14.1 Conceptos clave

```python
def ejemplo(*args, **kwargs):
    ...
```

```python
ejemplo(10, 20, 30, nombre="Ana", edad=20)
```

Internamente:

```python
args == (10, 20, 30)
kwargs == {"nombre": "Ana", "edad": 20}
```

Orden en la definición (simplificado):

`parametros_normales`, `*args`, `keyword_only`, `**kwargs`.


### 14.2 Explicación razonada

## Epígrafe 15 — Funciones como objetos

### 15.1 Conceptos clave

Las funciones **son objetos**. Por tanto pueden:

- asignarse a una variable,
- pasarse como argumento,
- devolverse desde otra función,
- guardarse en listas, dicts, etc.

```python
def sumar(a, b):
    return a + b

operacion = sumar          # guarda la FUNCIÓN
print(operacion(5, 3))     # 8

operacion = sumar(5, 3)    # guarda el RESULTADO (8)
```


### 15.2 Explicación razonada

`sumar` (sin paréntesis) es una referencia al objeto función. Tiene tipo, `.__name__`, `.__doc__`, se puede meter en un dict de operaciones.

`sumar(5, 3)` **invoca**. El valor de esa expresión es lo que `return` entregó.

Esta es la misma diferencia que `return incrementar` frente a `return incrementar(1)`.

```python
ops = {"+": sumar, "-": restar}
ops["+"](8, 5)   # 13
```

El dict no guarda números: guarda **capacidades**.

#### Cuándo

Calculadoras, callbacks, `key=` de `sorted`, estrategias.

#### Cuándo NO

`f = sumar(5, 3)` si luego querías `f(1, 2)`.


````python
# ── Práctica guiada 15 — la función es un objeto ─────────────────────────────
def sumar(a, b):
    return a + b

def restar(a, b):
    return a - b

operacion = sumar
print("operacion is sumar:", operacion is sumar)
print("operacion(5, 3) =", operacion(5, 3))

valor = sumar(5, 3)
print("valor =", valor, type(valor).__name__)

tabla = {"+": sumar, "-": restar}
print("tabla['-'](8, 5) =", tabla["-"](8, 5))
print("nombres:", [fn.__name__ for fn in tabla.values()])
````

## Epígrafe 16 — Pasar funciones como argumentos

### 16.1 Conceptos clave

```python
def sumar(a, b):
    return a + b

def restar(a, b):
    return a - b

def operar(funcion, a, b):
    return funcion(a, b)

operar(sumar, 8, 5)   # 13
operar(restar, 8, 5)  # 3
```

El parámetro `funcion` **no** es un string mágico: es un objeto callable. `operar` no sabe hacer sumas; **delega**.


### 16.2 Explicación razonada

`operar(sumar, 8, 5)` ata:

- `funcion →` objeto `sumar`
- `a → 8`, `b → 5`

Luego `funcion(a, b)` es exactamente `sumar(8, 5)`.

Si escribes `operar(sumar(), 8, 5)` evalúas `sumar()` **antes**, sin argumentos → error, o `None` si no tenía parámetros, y `funcion` deja de ser la función.

Este patrón es *inyección de comportamiento*: la misma plantilla (`aplicar a cada nota`, `ordenar`, `reintentar`) con distinta política.

`map` y `filter` (epígrafe 18) son eso mismo en la biblioteca estándar.

#### Cuándo

Callbacks, `key=`, estrategias, tests que inyectan un doble.

#### Cuándo NO

Un solo uso con una sola operación: llama directamente `sumar(8, 5)`.


````python
# ── Práctica guiada 16 — funciones como argumentos ───────────────────────────
def sumar(a, b):
    return a + b

def restar(a, b):
    return a - b

def multiplicar(a, b):
    return a * b

def operar(funcion, a, b):
    print(f"  ejecutando {funcion.__name__}")
    return funcion(a, b)

print(operar(sumar, 8, 5))
print(operar(restar, 8, 5))
print(operar(multiplicar, 8, 5))

def aplicar_a_notas(fn, notas):
    resultado = []
    for n in notas:
        resultado.append(fn(n))
    return resultado

print("redondeo:", aplicar_a_notas(round, [4.6, 7.2]))
````

## Epígrafe 17 — Devolver funciones

### 17.1 Conceptos clave

```python
def operar(orden):
    def incrementar(n):
        return n + 1
    def decrementar(n):
        return n - 1
    match orden:
        case "i":
            return incrementar
        case "d":
            return decrementar

var_funcion = operar("i")
print(var_funcion(6))   # 7
```

`operar` es una **fábrica** de funciones.


### 17.2 Explicación razonada, paso a paso

1. Se llama `operar("i")`.
2. Se definen `incrementar` y `decrementar` en ese marco.
3. `match` elige **devolver el objeto** `incrementar` (sin `()`).
4. Ese objeto se guarda en `var_funcion`.
5. `var_funcion(6)` ejecuta `incrementar(6)` → `7`.

También: `print(operar("i")(6))` — dos llamadas encadenadas.

#### `return incrementar` vs `return incrementar(1)`

| Código | Qué sale de `operar("i")` | `var(6)` |
|---|---|---|
| `return incrementar` | función | `7` |
| `return incrementar(1)` | `2` | `TypeError` |

#### Relación con closures

Si la interna usa variables de la externa, la función devuelta **recuerda** ese entorno (`nonlocal`, epígrafe 11).

#### Cuándo

Seleccionar política una vez y aplicarla muchas (`var_funcion` en un bucle). Configurar (`hacer_multiplicador(3)`).

#### Cuándo NO

Si siempre llamas una sola vez: `incrementar(6)` directo.


````python
# ── Práctica guiada 17 — devolver la función ─────────────────────────────────
def operar(orden):
    def incrementar(n):
        return n + 1

    def decrementar(n):
        return n - 1

    match orden:
        case "i":
            return incrementar
        case "d":
            return decrementar
        case _:
            return None

var_funcion = operar("i")
print("paso 1, objeto:", var_funcion)
print("paso 2, llamada:", var_funcion(6))
print("encadenado operar('d')(6) =", operar("d")(6))

def mal(orden):
    def incrementar(n):
        return n + 1
    match orden:
        case "i":
            return incrementar(1)

valor = mal("i")
print("mal devuelve:", valor, type(valor).__name__)
try:
    valor(6)
except TypeError as e:
    print("mal al llamar:", e)

def hacer_multiplicador(k):
    def mult(n):
        return n * k
    return mult

por3 = hacer_multiplicador(3)
print("por3(10) =", por3(10))
````

## Epígrafe 18 — `lambda`

### 18.1 Conceptos clave

```python
lambda parametros: expresion
```

Equivalencia:

```python
def doble(x):
    return x * 2

doble = lambda x: x * 2
```

| | `def` | `lambda` |
|---|---|---|
| Nombre | Obligatorio en el `def` | Anónima (puedes asignarla) |
| Cuerpo | Varias sentencias | **Una expresión** |
| `return` | Explícito | Implícito (el valor de la expresión) |
| Docstring | Sí | No idiomático |

`lambda` es un objeto función más: se asigna, se pasa, se devuelve.


### 18.2 Explicación razonada

`lambda` no es más rápida ni «más Python». Es **azúcar** para funciones de una expresión, típicas como `key=` de `sorted`, o callbacks minúsculos.

```python
alumnos = [
    {"nombre": "Ana", "nota": 8},
    {"nombre": "Luis", "nota": 5},
    {"nombre": "Pedro", "nota": 9},
]
ordenados = sorted(alumnos, key=lambda a: a["nota"])
```

`sorted` necesita una función que, dado un elemento, devuelva la clave de orden. La lambda **es** esa función. `sorted(alumnos, key=lambda a: a["nota"]())` estaría mal: `nota` no es callable.

`reverse=True` para de mayor a menor.

#### `map` y `filter` (sin convertir esto en un curso de funcional)

```python
list(map(lambda x: x * 2, [1, 2, 3]))       # [2, 4, 6]
list(filter(lambda x: x >= 5, [4, 5, 8]))   # [5, 8]
```

En Python moderno, las **comprensiones** suelen ser más claras: `[x * 2 for x in ...]` y `[x for x in ... if x >= 5]`. Conoce `map`/`filter` para leer código; no los fuerces.

#### Cuándo usar `lambda`

- `key=` de una línea.
- Argumento función que no merece un `def` ni un nombre.

#### Cuándo NO

- Cuerpo con `if`/`for`/`return` múltiples → `def`.
- Necesitas docstring o depurar con nombre claro.
- Lambda de 80 caracteres anidada: es «listo», no legible.

Asignar `doble = lambda x: x * 2` en un módulo suele ser peor estilo que `def doble`.


````python
# ── Práctica guiada 18 — lambda ──────────────────────────────────────────────
doble = lambda x: x * 2
cuadrado = lambda x: x ** 2
print(doble(5), cuadrado(5))

alumnos = [
    {"nombre": "Ana", "nota": 8},
    {"nombre": "Luis", "nota": 5},
    {"nombre": "Pedro", "nota": 9},
]
print("por nota asc:", sorted(alumnos, key=lambda a: a["nota"]))
print("por nota desc:", sorted(alumnos, key=lambda a: a["nota"], reverse=True))
print("por nombre:", sorted(alumnos, key=lambda a: a["nombre"]))

print("map:", list(map(lambda x: x * 2, [1, 2, 3])))
print("filter:", list(filter(lambda n: n >= 5, [4, 5, 8, 3])))
print("comprensión equivalente:", [n for n in [4, 5, 8, 3] if n >= 5])
````
